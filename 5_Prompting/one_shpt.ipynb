```python
example_input = "I absolutely loved the new restaurant downtown!"
example_output = """
<sentiment>
    <label>positive</label>
    <confidence>0.95</confidence>
</sentiment>
"""

system_prompt = f"""You analyze sentiment and respond only in XML using this structure:
<sentiment>
    <label>positive|negative|neutral</label>
    <confidence>0.0-1.0</confidence>
</sentiment>

For any other query just say I cannot do it.

Example input:
{example_input}
Example output:
{example_output}
"""

user_query = "The delivery was late and the food arrived cold."

input = f"""
Input: {user_query}
Output:
"""

print(input)
```

```python
print(system_prompt)
```

```python
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

client = Groq()

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_query}
    ]
)

print(response.choices[0].message.content)
```
