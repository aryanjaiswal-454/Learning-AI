# RAG (Retrieval-Augmented Generation)

## Definition
- **RAG** is the process of optimizing LLM output by referencing an authoritative external knowledge base before generating a response.
- It combines information retrieval with LLM generation.
- It extends LLM capabilities to specific domains or an organization's internal knowledge base without retraining the model.
- It improves the relevance, accuracy, and usefulness of LLM responses.

## Why RAG?
- Provides access to external and updated knowledge.
- Enables LLMs to answer questions using private or domain-specific documents.
- Reduces hallucinations by grounding answers in retrieved information.
- More cost-effective than retraining an LLM.

## RAG Workflow

1. **Document Loading:** Load data from PDFs, websites, documents, etc.
2. **Text Splitting:** Divide documents into smaller chunks.
3. **Embedding:** Convert text chunks into numerical vectors.
4. **Vector Storage:** Store embeddings in a vector database called as knowledge base.
5. **Retrieval:** Retrieve relevant chunks based on the user query.
6. **Generation:** Provide retrieved context and the query to the LLM to generate an answer.



## Key Components

- **Embeddings:** Convert text into numerical representations.
- **Vector Database:** Stores and searches embeddings based on similarity.
- **Retriever:** Finds relevant documents for a query.
- **LLM:** Generates the final response using retrieved context.

## RAG Architecture

Documents → Chunking → Embeddings → Vector Database

User Query → Retrieval → Relevant Context + Query → LLM → Answer

## Benefits
- Reduces hallucinations.
- Provides domain-specific knowledge.
- Supports updated external information.
- Avoids the need to retrain the LLM.
- Improves response relevance and accuracy.

## RAG Workflow / Pipeline

### 1. Data Ingestion Pipeline (Indexing)

- **Document Loading:** Collect data from sources such as PDFs, websites, HTML, Excel, CSV, and databases.
- **Parsing:** Extract meaningful text and content from the collected documents.
- **Chunking:** Split large documents into smaller chunks for efficient processing.
- **Embedding:** Convert text chunks into numerical vectors using an embedding model.
- **Vector Storage:** Store embeddings along with their text chunks in a Vector Database, which acts as the **Knowledge Base**.

### 2. Retrieval Pipeline (Query Processing)

- **User Query:** User submits a question.
- **Query Embedding:** Convert the user query into a vector using the same embedding model.
- **Similarity Search:** Search the Vector Database to find chunks most relevant to the query.
- **Retrieval:** Fetch the relevant chunks from the Knowledge Base.

### 3. Generation Pipeline

- **Context + Prompt:** Combine the retrieved information with the user's query and instructions.
- **LLM:** Pass the combined prompt to the LLM.
- **Output:** LLM generates a context-aware answer based on the retrieved information.

Documents → Parsing → Chunking → Embedding → Vector Database (Knowledge Base)

**Retrieval & Generation Pipeline:**

User Query → Query Embedding → Similarity Search → Relevant Chunks

Relevant Chunks + User Query → Prompt → LLM → Final Answer

- Provides access to external and updated knowledge.
