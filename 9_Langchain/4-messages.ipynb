## Messages in LangChain
Messages are the fundamental units of conversation in LangChain. They represent the inputs and outputs exchanged between a user, an LLM, and other components (like tools).

#### Key Components of a Message
1. Role: Identifies the type of message, such as system, human, ai, or tool.
2. Content: Contains the message data, such as text, images, audio, or documents.
3. Metadata: Optional information, such as response details, message IDs, token usage, and other model-specific information.

#### Text Prompts in LangChain
Text prompts are simple strings passed to an LLM for straightforward generation tasks. They are ideal when you don't need to maintain conversation history.

#### Use Text Prompts When
- You have a single, standalone request.
- You don't need conversation history.
- You want minimal code complexity.

```python
model.invoke("What is LangChain?")
```

#### Message Prompts in LangChain
Message prompts allow you to pass a list of message objects to a chat model. Each message has a role and content, enabling structured conversations and better control over model behavior.

#### Common Message Types
- SystemMessage: Defines the model's behavior, instructions, and role.
- HumanMessage: Represents input from the user.
- AIMessage: Represents the model's response and may include tool calls.
- ToolMessage: Contains the result returned by a tool after execution.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on Artificial Intelligence")
]

response = model.invoke(messages)
print(response.content)
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
system_msg = SystemMessage("""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Message Metadata
human_msg = HumanMessage(
    content="Hello!",
    name="alice",  # (metadata 1) Optional: identify different users
    id="msg_123",  # (metadata 2) Optional: unique identifier for tracing
)
```

```python
response=model.invoke([human_msg])
print(response)
print(response.usage_metadata)
```

```python
ai_msg = AIMessage("I'd be happy to help you with that question!")

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great!, What is 3*5+2?")
]

response= model.invoke(messages)
print(response,"\n\n",response.content)
```

```python
response.usage_metadata
```

```python
ai_message = AIMessage(
    content=[],
    tool_calls=[
        {
            "name":"get_weather",
            "args":{"location":"San Francisco"},
            "id":"call_123"
        }
    ]
)

weather_result = "Sunny, 72° F"
tool_message = ToolMessage(
    content = weather_result,
    tool_call_id = "call_123"
)

messages = [
    # SystemMessage()
    HumanMessage("What is the weather in San Francisco?"),
    ai_message,
    tool_message,
    # HumanMessage("What is the weather in London")
]

response=model.invoke(messages)
print(response.content)
```
