## MCP (Model Context Protocol)

- **MCP** is an open protocol that allows AI applications to connect with external tools, data sources, and services.
- It standardizes communication between AI applications (MCP Clients) and external services (MCP Servers).

### Key Components

- **MCP Host:** AI application that manages MCP connections (e.g., Claude Desktop).
- **MCP Client:** Connects the host to an MCP server.
- **MCP Server:** Exposes external tools, resources, and prompts.

### Main MCP Primitives

- **Tools:** Functions the AI can call to perform tasks.
- **Resources:** Provide data or context to the AI.
- **Prompts:** Reusable prompt templates.

### Working

`AI Application → MCP Client → MCP Server → External Tools / Data`

### Benefits

- Standardized integration with external services.
- Reusable tools across different AI applications.
- Reduces the need for custom integrations.
- Supports dynamic discovery of available tools and resources.

## Transport in MCP (Model Context Protocol)

- **Transport** defines how MCP Clients and MCP Servers communicate and exchange messages.
- It handles the communication channel between the client and server.

### Types of Transport

- **STDIO:** Tells the server to communicate through standard input/output (stdin and stdout) like in CLI, commonly used for local MCP servers. Suitable for testing locally.
- **Streamable HTTP:** Allows communication with remote MCP servers over HTTP using streaming capabilities.

**Important:** Transport is responsible for message delivery, not for defining what tools or resources do.
